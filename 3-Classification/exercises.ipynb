```python
from sklearn.datasets import fetch_openml

mnist = fetch_openml('mnist_784',as_frame=False)

X , y = mnist.data , mnist.target

X.shape
```

```python
X_train , X_test ,y_train,y_test = X[:60000],X[60000:],y[:60000],y[60000:]
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn_clf = KNeighborsClassifier(n_neighbors=3,weights="distance")
knn_clf.fit(X_train,y_train)
```

```python
from sklearn.model_selection import cross_val_score

cross_val_score(knn_clf,X_train,y_train,cv=3,scoring="accuracy")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV

full_pipeline = Pipeline ([
    ("knn",knn_clf)
])

param_grid=[{
    'knn__weights':['distance'],'knn__n_neighbors':[1,2,3,5,6,7]
},
{
    'knn__weights':['uniform'],'knn__n_neighbors':[1,2,3,5,6,7]
}]

grid_search = GridSearchCV(full_pipeline,param_grid,scoring="accuracy",cv=3)
grid_search.fit(X_train,y_train)
```

```python
grid_search.best_params_
```

```python
tuned_accuracy = grid_search.score(X_test,y_test)
tuned_accuracy
```

```python
from scipy.ndimage import shift

def shiftt(image,dx,dy):
    image=image.reshape(28,28)
    shifted_image = shift(image,[dx,dy],cval=0,mode="constant")
    return shifted_image.reshape([-1])

X_train_augmented = [image for image in X_train]
y_train_augmented = [label for label in y_train]


for x_shifted,y_shifted in zip(mnist.data,mnist.target):
    X_right=shiftt(x_shifted,0,1)
    X_left=shiftt(x_shifted,0,-1)
    X_down=shiftt(x_shifted,1,0)
    X_up=shiftt(x_shifted,-1,0)
    X_train_augmented.append(X_right)
    y_train_augmented.append(y_shifted)
    X_train_augmented.append(X_left)
    y_train_augmented.append(y_shifted)
    X_train_augmented.append(X_down)
    y_train_augmented.append(y_shifted)
    X_train_augmented.append(X_up)
    y_train_augmented.append(y_shifted)


    
    

    
```

```python
import numpy as np

X_train_augmented = np.array(X_train_augmented)
y_train_augmented = np.array(y_train_augmented)
```

```python

shuffeld_index=np.random.permutation(len(X_train_augmented))
X_train_augmented = X_train_augmented[shuffeld_index]
y_train_augmented = y_train_augmented[shuffeld_index]
```

```python
import numpy as np
import pandas as pd
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.preprocessing import OneHotEncoder
from sklearn.utils.validation import check_is_fitted

class TicketEncoder(BaseEstimator, TransformerMixin):
    def __init__(self):
        self.encoder = OneHotEncoder(sparse_output=False, handle_unknown="ignore")

    def fit(self, X, y=None):
        ticket = X.iloc[:, 0].fillna('another_ignore').astype(str)
        cat = ticket.str.split(' ').str[0]
        self.encoder.fit(cat.to_frame())
        return self

    def transform(self, X):
        ticket = X.iloc[:, 0].fillna('another_ignore').astype(str)

        cat = ticket.str.split(' ').str[0]
        num = pd.to_numeric(
            ticket.str.split(' ').str[-1],
            errors='coerce'
        ).fillna(0).values

        encoded = self.encoder.transform(cat.to_frame())
        return encoded * num[:, None]

    def get_names_out(self, input_features=None):
        return np.array(["Ticket"])
```

```python
import re
import numpy as np
import pandas as pd
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.preprocessing import OneHotEncoder
from sklearn.utils.validation import check_is_fitted

class CabinEncoder(BaseEstimator, TransformerMixin):
    def __init__(self):
        self.encoder = OneHotEncoder(sparse_output=False, handle_unknown="ignore")

    def fit(self, X, y=None):
        cabin = X.iloc[:, 0].fillna('another_ignore').astype(str)
        cat = cabin.str.replace(r'\d+\.?\d*', '', regex=True)
        self.encoder.fit(cat.to_frame())
        return self

    def transform(self, X):
        cabin = X.iloc[:, 0].fillna('another_ignore').astype(str)
        cat = cabin.str.replace(r'\d+\.?\d*', '', regex=True)
        num = pd.to_numeric(
            cabin.str.extract(r'(\d+\.?\d*)')[0],
            errors='coerce'
        ).fillna(0).values

        encoded = self.encoder.transform(cat.to_frame())
        return encoded * num[:, None]

    def get_names_out(self, input_features=None):
        return np.array(["Cabin"])
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler,OneHotEncoder

num_attribs = ["Pclass","Age","SibSp","Parch","Fare"]
cat_attribs = ["Embarked","Sex"]


cat_pipeline = make_pipeline(
    SimpleImputer(strategy = "most_frequent"),
    OneHotEncoder(handle_unknown="ignore")
)

num_pipeline = make_pipeline(
    SimpleImputer(strategy="median"),
    StandardScaler()
)

cabin_pipeline = make_pipeline(
    CabinEncoder(),
)

ticket_pipeline = make_pipeline(
    TicketEncoder(),
)

preprocessing = ColumnTransformer([
    ("num",num_pipeline,num_attribs),
    ("cat",cat_pipeline,cat_attribs),
    ("cabin",cabin_pipeline,["Cabin"]),
    ("Ticket",ticket_pipeline,["Ticket"])
])
```

```python
from sklearn.ensemble import RandomForestClassifier

forest_clf = make_pipeline(preprocessing,RandomForestClassifier(random_state=42))
forest_clf.fit(titanic,titanic_labels)
```

```python
from sklearn.model_selection import cross_val_score

forest_scores = cross_val_score(forest_clf, titanic, titanic_labels, cv=10)
forest_scores.mean()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler,OneHotEncoder

num_attribs = ["Pclass","Age","SibSp","Parch","Fare","family"]
cat_attribs = ["Embarked","Sex"]


cat_pipeline = make_pipeline(
    SimpleImputer(strategy = "most_frequent"),
    OneHotEncoder(handle_unknown="ignore")
)

num_pipeline = make_pipeline(
    SimpleImputer(strategy="median"),
    StandardScaler()
)

cabin_pipeline = make_pipeline(
    CabinEncoder(),
)

ticket_pipeline = make_pipeline(
    TicketEncoder(),
)

preprocessing_svm = ColumnTransformer([
    ("num",num_pipeline,num_attribs),
    ("cat",cat_pipeline,cat_attribs),
])
```

```python
from sklearn.svm import SVC

svm_clf = make_pipeline(preprocessing_svm,SVC(gamma='auto'))
svm_clf.fit(titanic,titanic_labels)
```

```python
svm_scores = cross_val_score(svm_clf,titanic,titanic_labels,cv=10)
svm_scores.mean()
```
